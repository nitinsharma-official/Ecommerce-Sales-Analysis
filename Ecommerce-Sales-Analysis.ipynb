```python
# E-Commerce Sales Analysis
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
print("Libraries imported successfully")
```

```python
data = {
    "Order_ID": [1001,1002,1003,1004,1005,1006,1007,1008,1009,1010,
                 1011,1012,1013,1014,1015,1016,1017,1018,1019,1020],
    "Product": ["Laptop","Mobile","Headphones","Keyboard","Mouse",
                "Laptop","Mobile","Monitor","Headphones","Tablet",
                "Keyboard","Mouse","Laptop","Mobile","Monitor",
                "Tablet","Headphones","Keyboard","Mouse","Laptop"],
    "Category": ["Electronics","Electronics","Accessories","Accessories","Accessories",
                 "Electronics","Electronics","Electronics","Accessories","Electronics",
                 "Accessories","Accessories","Electronics","Electronics","Electronics",
                 "Electronics","Accessories","Accessories","Accessories","Electronics"],
    "Quantity": [2,3,5,4,6,1,4,2,5,3,7,8,1,2,3,2,6,5,9,2],
    "Price": [50000,25000,2000,1500,800,
              55000,24000,18000,2200,30000,
              1400,700,52000,26000,17500,
              28000,2100,1600,750,54000],
    "Discount": [10,5,0,10,5,15,10,5,0,10,
                 5,10,20,5,10,15,0,5,10,10],
    "Rating": [4.5,4.2,4.8,4.1,4.0,
               4.7,4.3,4.6,4.4,4.5,
               4.1,3.9,4.9,4.6,4.2,
               4.7,4.3,4.0,4.1,4.8]
}
df = pd.DataFrame(data)
print(df)
```

```python
print("Shape:")
print(df.shape)
print("\nColumns:")
print(df.columns)
print("\nData Types:")
print(df.dtypes)
print("\nFirst 5 rows:")
print(df.head())
```

```python
print(df.isnull().sum()) # missing values check
```

```python
# Gross Sales = Quantity × Price
df["Gross_Sales"] = df["Quantity"] * df["Price"]
print(df[["Order_ID", "Product", "Quantity", "Price", "Gross_Sales"]])
```

```python
# Discount Amount = Gross Sales × Discount / 100
df["Discount_Amount"] = df["Gross_Sales"] * df["Discount"] / 100
print(df[["Order_ID", "Gross_Sales", "Discount", "Discount_Amount"]])
```

```python
# Revenue = Gross Sales - Discount Amount
df["Revenue"] = df["Gross_Sales"] - df["Discount_Amount"]
print(df[["Order_ID", "Gross_Sales", "Discount_Amount", "Revenue"]])
```

```python
# Total Revenue 
total_revenue = df["Revenue"].sum()
print("Total Revenue:", total_revenue)
```

```python
# Average Revenue
average_revenue = df["Revenue"].mean()
print("Average Order Revenue:", average_revenue)
```

```python
# Highest Revenue 
highest_revenue = df["Revenue"].max()
print("Highest Revenue:", highest_revenue)
```

```python
best_order = df.loc[df["Revenue"].idxmax()]
print(best_order)
```

```python
# Product-wise sales
product_sales = df.groupby("Product")["Revenue"].sum()
print(product_sales)
```

```python
# Category-wise Sales
category_sales = df.groupby("Category")["Revenue"].sum()
print(category_sales)
```

```python
# Best Product
best_product = product_sales.idxmax()
best_product_revenue = product_sales.max()
print("Best Product:", best_product)
print("Revenue:", best_product_revenue)
```

```python
# Rating Analysis
high_rated = df[df["Rating"] >= 4.5]
print(high_rated)
```

```python
# Sort Revenue
sorted_sales = df.sort_values("Revenue", ascending=False)
print(sorted_sales)
```

```python
# Top 5 Revenue
top_5 = df.sort_values("Revenue", ascending=False).head(5)
print(top_5)
```

```python
# Visualization 
# Revenue by Product
import matplotlib.pyplot as plt
product_revenue = df.groupby("Product")["Revenue"].sum()
plt.figure(figsize=(10, 5))
product_revenue.plot(
    kind="bar",
    color=["#3498DB", "#E74C3C", "#2ECC71", "#F39C12", "#9B59B6", "#1ABC9C", "#E67E22"]
)
plt.title("Revenue by Product", fontsize=16, fontweight="bold")
plt.xlabel("Product")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.grid(axis="y", alpha=0.3)
plt.show()
```

```python
# Category Revenue Chart
category_revenue = df.groupby("Category")["Revenue"].sum()
plt.figure(figsize=(8, 5))
category_revenue.plot(
    kind="bar",
    color=["#3498DB", "#E74C3C"]
)
plt.title("Revenue by Category", fontsize=16, fontweight="bold")
plt.xlabel("Category")
plt.ylabel("Revenue")
plt.grid(axis="y", alpha=0.3)
plt.show()
```

```python
# Rating Distribution
plt.figure(figsize=(8, 5))
plt.hist(
    df["Rating"],
    bins=5,
    color="#9B59B6",
    edgecolor="black"
)
plt.title("Customer Rating Distribution", fontsize=16, fontweight="bold")
plt.xlabel("Rating")
plt.ylabel("Number of Orders")
plt.grid(axis="y", alpha=0.3)
plt.show()
```

```python
# Quantity Sold by Product
product_quantity = df.groupby("Product")["Quantity"].sum()
plt.figure(figsize=(10, 5))
product_quantity.plot(
    kind="bar",
    color=["#1ABC9C", "#9B59B6", "#F1C40F", "#E67E22", "#2ECC71", "#3498DB", "#E74C3C"]
)
plt.title("Quantity Sold by Product", fontsize=16, fontweight="bold")
plt.xlabel("Product")
plt.ylabel("Quantity Sold")
plt.xticks(rotation=45)
plt.grid(axis="y", alpha=0.3)
plt.show()
```

```python
# Final Business Insights
print("========== BUSINESS INSIGHTS ==========")
print("Total Revenue:", total_revenue)
print("Average Order Revenue:", average_revenue)
print("Best Product:", best_product)
print("Best Product Revenue:", best_product_revenue)
print("Highest Revenue Order:")
print(best_order)
```
